# Resumo dos Dados

Dicionário das métricas do Oracle’s Elixir, organizado por categoria.

> **Abreviações:** CS = tropas e monstros abatidos; wards = sentinelas; pós-15 = período após os 15 minutos de partida.
>
> As métricas podem representar jogadores, campeões ou equipes, dependendo da tabela consultada.

## 1. Identificação

| Coluna | Descrição |
|---|---|
| `Champion` | Nome do campeão. |
| `Event` | Nome do evento ou campeonato. |
| `Player` | Nome do jogador dentro do jogo. |
| `Pos` | Posição ou função. |
| `Team` | Nome da equipe. |
| `Rank` | Posição oficial no ranking. |
| `LP` | Pontos de classificação (*Ladder Points*). |

## 2. Partidas e Resultados

| Coluna | Descrição |
|---|---|
| `GP` | Total de partidas disputadas. |
| `W` / `Wins` | Total de vitórias. |
| `L` / `Losses` | Total de derrotas. |
| `W%` | Porcentagem de vitórias. |
| `AGT` | Duração média das partidas, em minutos. |

## 3. Abates, Mortes e Assistências

| Coluna | Descrição |
|---|---|
| `K` | Total de abates. |
| `D` | Total de mortes. |
| `A` | Total de assistências. |
| `KPG` | Média de abates por partida. |
| `DPG` | Média de mortes por partida. |
| `APG` | Média de assistências por partida. |
| `KD` | Razão entre abates e mortes. |
| `KDA` | Razão entre a soma de abates e assistências e o total de mortes. |
| `KP` | Participação em abates: porcentagem dos abates da equipe em que o jogador conseguiu um abate ou uma assistência. |
| `KS%` | Porcentagem dos abates da equipe realizados pelo jogador. |
| `DTH%` | Participação média nas mortes da equipe. |
| `CKPM` | Média de abates combinados por minuto: abates da equipe mais abates do adversário. |
| `FB%` | Taxa de primeiro abate (*First Blood*). Para jogadores e campeões, considera a participação por abate ou assistência. |
| `FBV%` | Porcentagem de partidas em que o jogador ou campeão foi a vítima do primeiro abate. |

## 4. Dano e Controle de Grupo

| Coluna | Descrição |
|---|---|
| `DPM` | Média de dano causado a campeões por minuto. |
| `DMG%` | Participação média no dano total da equipe contra campeões. |
| `D%P15` / `DMG%P15` | Participação média no dano da equipe contra campeões após os 15 minutos. |
| `CCPM` | Controle de grupo aplicado a campeões por minuto. |

## 5. Farm e Controle de Recursos

| Coluna | Descrição |
|---|---|
| `CSPM` | Média de tropas e monstros abatidos por minuto. |
| `CSD10` | Diferença média de CS aos 10 minutos. |

| `CSD20` | Diferença média de CS aos 20 minutos. |
| `CS%P15` | Participação média no CS total da equipe após os 15 minutos. |
| `JNG%` | Controle da selva: participação média no CS total de selva da partida. |
| `LNE%` | Controle de rotas: participação média no CS total de tropas das rotas da partida. |

## 6. Ouro e Experiência

# Recuperação de Dados, Limpeza e Tratamento.

## Separação entre jogadores e equipes

O dataset reúne estatísticas individuais e coletivas. Nesta etapa,
separamos os registros pela coluna `position`:

- **Jogadores:** posições `top`, `jng`, `mid`, `bot` e `sup`.
- **Equipes:** posição `team`.

Criamos dois DataFrames independentes, preservando o DataFrame original.
Cada linha continua representando um jogador ou uma equipe em uma partida.

```python
posicoes_jogadores = ["top", "jng", "mid", "bot", "sup"]

df_jogadores = df.loc[
    df["position"].isin(posicoes_jogadores)
].copy()

df_times = df.loc[
    df["position"].eq("team")
].copy()

print(f"Registros originais: {len(df)}")
print(f"Registros de jogadores: {len(df_jogadores)}")
print(f"Registros de equipes: {len(df_times)}")
print(f"Registros não classificados: {len(df) - len(df_jogadores) - len(df_times)}")
```

## Seleção das principais ligas

Selecionamos registros das seis ligas regionais de primeira divisão:

- **LCK:** Coreia do Sul.
- **LPL:** China.
- **LEC:** Europa, Oriente Médio e África.
- **LCS:** América do Norte.
- **LCP:** Ásia-Pacífico.
- **CBLOL:** Brasil.

Aplicamos o mesmo filtro aos dados de jogadores e equipes, criando
DataFrames específicos para esse recorte.

A seleção exclui competições secundárias e torneios internacionais.
Pertencer à primeira divisão não significa que todas as ligas tenham
o mesmo nível competitivo.

```python
ligas_principais = ["LCK", "LPL", "LEC", "LCS", "LCP", "CBLOL"]

df_jogadores_principais = df_jogadores.loc[
    df_jogadores["league"].isin(ligas_principais)
].copy()

df_times_principais = df_times.loc[
    df_times["league"].isin(ligas_principais)
].copy()

resumo_ligas = pd.DataFrame({
    "registros_jogadores": df_jogadores_principais.groupby("league").size(),
    "registros_times": df_times_principais.groupby("league").size(),
}).sort_index()

resumo_ligas
```

## Seleção das colunas dos jogadores

Selecionamos as colunas necessárias para identificar os jogadores,
verificar a qualidade dos registros e analisar seu desempenho.

As colunas são organizadas em três grupos:

- **Contexto:** identificação, liga, equipe, posição, campeão e período.
  Serão utilizadas para filtragem e interpretação, sem entrar diretamente
  no cálculo dos agrupamentos.
- **Preparação:** duração, abates, mortes, assistências e abates da equipe.
  Permitem calcular taxas por minuto e participação em abates.
- **Métricas candidatas:** dano, recursos, visão e vantagens aos 15 minutos.

As demais colunas ficam fora desta base de trabalho. A seleção final das
variáveis do modelo será revisada após analisar valores ausentes,
redundância e distribuição das métricas.

Os registros continuam separados por jogador e partida.

```python
# Identificação, qualidade e contexto — fora das entradas do modelo.
colunas_contexto = [
    "gameid",
    "datacompleteness",
    "league",
    "split",
    "playoffs",
    "date",
    "patch",
    "side",
    "position",
    "playerid",
    "playername",
    "teamid",
    "teamname",
    "champion",
    "result",
]

# Dados para calcular métricas posteriormente.
colunas_preparacao = [
    "gamelength",
    "kills",
    "deaths",
    "assists",
    "teamkills",
]

# Métricas inicialmente consideradas para caracterizar os jogadores.
colunas_metricas = [
    "dpm",
    "damageshare",
    "earnedgoldshare",
    "cspm",
    "wpm",
    "wcpm",
    "golddiffat15",
    "xpdiffat15",
    "csdiffat15",
]

colunas_manter = (
    colunas_contexto
    + colunas_preparacao
    + colunas_metricas
)

colunas_excluidas = [
    coluna
    for coluna in df_jogadores_principais.columns
    if coluna not in colunas_manter
]

df_jogadores_selecionado = df_jogadores_principais[
    colunas_manter
].copy()

print(f"Colunas mantidas: {len(colunas_manter)}")
print(f"Colunas excluídas deste recorte: {len(colunas_excluidas)}")

df_jogadores_selecionado.head()
```

## Limpeza e tratamento dos dados

Criamos uma cópia da base selecionada para padronizar os tipos das
colunas, remover espaços extras em textos e verificar duplicatas.

Aplicamos os seguintes tratamentos:

- Conversão de identificadores e categorias para texto.
- Conversão da data para o tipo datetime.
- Conversão das medidas de desempenho para tipos numéricos.
- Preenchimento de `split` ausente com "Não informado".
- Remoção apenas de linhas totalmente idênticas, caso existam.
- Identificação dos registros sem métricas completas aos 15 minutos.

Preservamos os registros classificados como `partial`, pois as métricas
disponíveis ainda podem ser utilizadas. Valores numéricos ausentes
permanecem ausentes, sem substituição por zero.

Valores extremos não são removidos automaticamente: podem representar
atuações reais e serão avaliados na análise exploratória.

```python
df_jogadores_limpo = df_jogadores_selecionado.copy()

# 1. Padronização de textos.
colunas_texto = [
    "gameid", "datacompleteness", "league", "split",
    "patch", "side", "position", "playerid", "playername",
    "teamid", "teamname", "champion",
]

for coluna in colunas_texto:
    df_jogadores_limpo[coluna] = (
        df_jogadores_limpo[coluna]
        .astype("string")
        .str.strip()
        .replace("", pd.NA)
    )

# 2. Conversão das datas e medidas.
df_jogadores_limpo["date"] = pd.to_datetime(
    df_jogadores_limpo["date"],
    errors="raise",
)

colunas_numericas = [
    "playoffs", "result",
    *colunas_preparacao,
    *colunas_metricas,
]

for coluna in colunas_numericas:
    df_jogadores_limpo[coluna] = pd.to_numeric(
        df_jogadores_limpo[coluna],
        errors="raise",
    )

# 3. Categoria ausente, sem inventar a etapa da competição.
df_jogadores_limpo["split"] = (
    df_jogadores_limpo["split"].fillna("Não informado")
)

# 4. Remoção apenas de duplicatas exatas.
quantidade_antes = len(df_jogadores_limpo)

df_jogadores_limpo = (
    df_jogadores_limpo
    .drop_duplicates()
    .reset_index(drop=True)
)

print(
    "Duplicatas exatas removidas:",
    quantidade_antes - len(df_jogadores_limpo),
)

# 5. Sinalização da disponibilidade dos dados aos 15 minutos.
colunas_15min = [
    "golddiffat15",
    "xpdiffat15",
    "csdiffat15",
]

df_jogadores_limpo["dados_15min_completos"] = (
    df_jogadores_limpo[colunas_15min].notna().all(axis=1)
)

# 6. Verificação de possíveis registros conflitantes.
chave_registro = ["gameid", "side", "position"]

print(
    "Registros com chave repetida:",
    df_jogadores_limpo.duplicated(
        subset=chave_registro,
        keep=False,
    ).sum(),
)

# 7. Resumo dos valores ainda ausentes.
resumo_nulos = pd.DataFrame({
    "quantidade": df_jogadores_limpo.isna().sum(),
    "percentual": (
        df_jogadores_limpo.isna().mean().mul(100).round(2)
    ),
})

resumo_nulos.loc[resumo_nulos["quantidade"] > 0]
```

### Relações entre as métricas

Calculamos a correlação de Spearman entre as métricas dentro de uma posição.
Isso reduz a influência das diferenças entre funções na comparação.

Correlações próximas de 1 ou -1 indicam associações monotônicas fortes.
Variáveis fortemente relacionadas podem dar peso repetido a uma mesma
dimensão no agrupamento.

As correlações utilizam os pares de valores disponíveis. Valores ausentes
não são substituídos por zero. Correlação não demonstra causalidade.

```python
# Altere para "top", "jng", "bot" ou "sup" para explorar outras posições.
posicao_analisada = "mid"

dados_posicao = df_jogadores_limpo.loc[
    df_jogadores_limpo["position"].eq(posicao_analisada),
    colunas_metricas,
]

correlacoes = dados_posicao.corr(method="spearman")

plt.figure(figsize=(11, 8))

sns.heatmap(
    correlacoes,
    annot=True,
    fmt=".2f",
    cmap="vlag",
    vmin=-1,
    vmax=1,
    center=0,
    square=True,
)

plt.title(f"Correlação entre métricas — {posicao_analisada}")
plt.tight_layout()
plt.show()
```

### Cobertura das métricas aos 15 minutos

Analisamos a disponibilidade conjunta das diferenças de ouro, experiência
e CS aos 15 minutos.

Calculamos a quantidade e o percentual de registros completos por liga
e por jogador, mantendo a posição separada.

Essa verificação permite identificar se os dados ausentes estão concentrados
em determinadas ligas ou jogadores e avaliar o impacto de uma possível
exclusão desses registros.

A cobertura mede disponibilidade dos dados, não desempenho.

```python
colunas_15min = [
    "golddiffat15",
    "xpdiffat15",
    "csdiffat15",
]

# Atualiza o indicador usando as três métricas.
df_jogadores_limpo["dados_15min_completos"] = (
    df_jogadores_limpo[colunas_15min].notna().all(axis=1)
)

cobertura_ligas = (
    df_jogadores_limpo
    .groupby("league")
    .agg(
        registros=("gameid", "size"),
        registros_completos=("dados_15min_completos", "sum"),
    )
)

cobertura_ligas["registros_incompletos"] = (
    cobertura_ligas["registros"]
    - cobertura_ligas["registros_completos"]
)

cobertura_ligas["cobertura_pct"] = (
    cobertura_ligas["registros_completos"]
    .div(cobertura_ligas["registros"])
    .mul(100)
)

cobertura_ligas = cobertura_ligas.sort_values("cobertura_pct")

cobertura_ligas.round(2)
```

#### Cobertura por jogador e posição

Detalhamos a disponibilidade por jogador, liga e posição. A separação por
liga evita ocultar diferenças de cobertura quando um jogador atua em mais
de uma competição.

Exibimos primeiro os perfis com menor cobertura, incluindo quantas partidas
possuem as três métricas disponíveis.

```python
cobertura_jogadores = (
    df_jogadores_limpo
    .groupby(["playerid", "league", "position"])
    .agg(
        jogador=("playername", "first"),
        partidas=("gameid", "nunique"),
        partidas_completas=("dados_15min_completos", "sum"),
    )
    .reset_index()
)

cobertura_jogadores["partidas_incompletas"] = (
    cobertura_jogadores["partidas"]
    - cobertura_jogadores["partidas_completas"]
)

cobertura_jogadores["cobertura_pct"] = (
    cobertura_jogadores["partidas_completas"]
    .div(cobertura_jogadores["partidas"])
    .mul(100)
)

jogadores_afetados = (
    cobertura_jogadores
    .loc[lambda tabela: tabela["partidas_incompletas"] > 0]
    .sort_values(
        ["cobertura_pct", "partidas_incompletas"],
        ascending=[True, False],
    )
)

print(
    "Jogadores únicos com alguma partida incompleta:",
    jogadores_afetados["playerid"].nunique(),
)

print(
    "Perfis jogador–liga–posição sem nenhuma partida completa:",
    cobertura_jogadores["partidas_completas"].eq(0).sum(),
)

jogadores_afetados[
    [
        "jogador",
        "league",
        "position",
        "partidas",
        "partidas_completas",
        "partidas_incompletas",
        "cobertura_pct",
    ]
].head(30).round(2)
```

### Construção dos perfis por jogador e posição

Agregamos as partidas para representar cada jogador em cada posição.

Calculamos:

- Média das métricas por partida, atribuindo o mesmo peso a cada partida.
- Desvio-padrão de algumas métricas para descrever a variação entre atuações.
- Quantidade de partidas e observações disponíveis aos 15 minutos.
- Abates, mortes e assistências por minuto.
- Participação em abates da equipe.

A duração original está em segundos e é convertida para minutos.

Valores ausentes são ignorados no cálculo das médias. Quando não existe
nenhuma observação disponível, a média permanece ausente. O desvio-padrão
também permanece ausente quando há menos de duas observações.

Os perfis resumem o período analisado e podem reunir diferentes equipes,
campeões e patches. A variabilidade observada também reflete esses contextos.

```python
df_perfis_base = df_jogadores_limpo.copy()

# Duração em minutos, protegendo contra durações inválidas.
minutos = df_perfis_base["gamelength"].div(60)
minutos = minutos.where(minutos > 0)

# Taxas calculadas para cada partida.
df_perfis_base["kills_pm"] = df_perfis_base["kills"] / minutos
df_perfis_base["deaths_pm"] = df_perfis_base["deaths"] / minutos
df_perfis_base["assists_pm"] = df_perfis_base["assists"] / minutos

# Se a equipe não fez abates, a participação fica ausente.
abates_equipe = df_perfis_base["teamkills"].where(
    df_perfis_base["teamkills"] > 0
)

df_perfis_base["participacao_abates"] = (
    df_perfis_base["kills"] + df_perfis_base["assists"]
) / abates_equipe

metricas_perfil = [
    *colunas_metricas,
    "kills_pm",
    "deaths_pm",
    "assists_pm",
    "participacao_abates",
]

metricas_variabilidade = [
    "dpm",
    "cspm",
    "deaths_pm",
    "golddiffat15",
]

# Ordenação para guardar o nome mais recente de cada jogador.
grupos = (
    df_perfis_base
    .sort_values("date")
    .groupby(["playerid", "position"])
)

contexto_perfis = grupos.agg(
    jogador=("playername", "last"),
    ligas=("league", lambda x: ", ".join(sorted(x.dropna().unique()))),
    equipes=("teamname", lambda x: ", ".join(sorted(x.dropna().unique()))),
    partidas=("gameid", "nunique"),
    primeira_partida=("date", "min"),
    ultima_partida=("date", "max"),
    partidas_15min_completas=("dados_15min_completos", "sum"),
)

medias_perfis = grupos[metricas_perfil].mean().add_suffix("_media")

desvios_perfis = (
    grupos[metricas_variabilidade]
    .std()
    .add_suffix("_desvio")
)

# Registra quantas observações sustentam cada média.
contagens_perfis = grupos[metricas_perfil].count().add_suffix("_n")

df_perfis = (
    contexto_perfis
    .join(medias_perfis)
    .join(desvios_perfis)
    .join(contagens_perfis)
    .reset_index()
)

df_perfis["cobertura_15min_pct"] = (
    df_perfis["partidas_15min_completas"]
    .div(df_perfis["partidas"])
    .mul(100)
)

print(f"Jogadores únicos: {df_perfis['playerid'].nunique()}")
print(f"Perfis jogador–posição: {len(df_perfis)}")

df_perfis.head()
```

### Correlações entre os perfis agregados

Calculamos a correlação de Spearman entre as médias dos jogadores,
separadamente para cada posição.

Nesta comparação, consideramos perfis com pelo menos 20 partidas.
Esse limite é exploratório e não garante, sozinho, estabilidade das médias.

Cada jogador tem uma observação por posição, independentemente de quantas
partidas disputou. Assim, jogadores com mais partidas não contribuem com
mais linhas para a correlação.

Utilizamos inicialmente métricas sem dependência dos dados aos 15 minutos.
Buscamos identificar variáveis relacionadas que possam dar peso repetido
a uma dimensão do desempenho no clustering.

Correlações elevadas não levam à exclusão automática de uma variável:
também devemos considerar o significado de cada métrica.

```python
minimo_partidas_analise = 20

metricas_candidatas = [
    "dpm_media",
    "damageshare_media",
    "earnedgoldshare_media",
    "cspm_media",
    "wpm_media",
    "wcpm_media",
    "kills_pm_media",
    "deaths_pm_media",
    "assists_pm_media",
    "participacao_abates_media",
]

perfis_analise = df_perfis.loc[
    df_perfis["partidas"].ge(minimo_partidas_analise)
].copy()

correlacoes_por_posicao = {}
pares_correlacionados = []

for posicao in ["top", "jng", "mid", "bot", "sup"]:
    dados = perfis_analise.loc[
        perfis_analise["position"].eq(posicao),
        metricas_candidatas,
    ]

    matriz = dados.corr(method="spearman")
    correlacoes_por_posicao[posicao] = matriz

    plt.figure(figsize=(11, 9))

    sns.heatmap(
        matriz,
        annot=True,
        fmt=".2f",
        cmap="vlag",
        vmin=-1,
        vmax=1,
        center=0,
        square=True,
        annot_kws={"size": 8},
    )

    plt.title(
        f"Correlação entre médias — {posicao}\n"
        f"{len(dados)} perfis com pelo menos "
        f"{minimo_partidas_analise} partidas"
    )
    plt.tight_layout()
    plt.show()

    # Examina cada par uma única vez, sem a diagonal.
    for i, metrica_a in enumerate(metricas_candidatas):
        for metrica_b in metricas_candidatas[i + 1:]:
            valor = matriz.loc[metrica_a, metrica_b]

            if pd.notna(valor) and abs(valor) >= 0.75:
                pares_correlacionados.append({
                    "posicao": posicao,
                    "metrica_a": metrica_a,
                    "metrica_b": metrica_b,
                    "correlacao": valor,
                    "perfis_disponiveis": (
                        dados[[metrica_a, metrica_b]]
                        .dropna()
                        .shape[0]
                    ),
                })

resumo_correlacoes = pd.DataFrame(
    pares_correlacionados,
    columns=[
        "posicao",
        "metrica_a",
        "metrica_b",
        "correlacao",
        "perfis_disponiveis",
    ],
)

resumo_correlacoes = (
    resumo_correlacoes
    .assign(intensidade=lambda tabela: tabela["correlacao"].abs())
    .sort_values("intensidade", ascending=False)
    .drop(columns="intensidade")
    .reset_index(drop=True)
)

resumo_correlacoes.round(3)
```

### Seleção das entradas do modelo

Selecionamos oito métricas de dano, recursos, visão e participação
em combate para os perfis com pelo menos 20 partidas.

Criamos uma matriz para cada posição. Identificadores, ligas, equipes,
contagens e indicadores de cobertura ficam fora das entradas.

Preservamos os índices para relacionar posteriormente cada agrupamento
ao jogador correspondente. As variáveis ainda não estão padronizadas.

```python
minimo_partidas = 20

features_modelo = [
    "dpm_media",
    "earnedgoldshare_media",
    "cspm_media",
    "wpm_media",
    "wcpm_media",
    "deaths_pm_media",
    "assists_pm_media",
    "participacao_abates_media",
]

df_modelo = df_perfis.loc[
    df_perfis["partidas"].ge(minimo_partidas)
].copy()

X_por_posicao = {
    posicao: grupo[features_modelo].copy()
    for posicao, grupo in df_modelo.groupby("position")
}

resumo_entradas = pd.DataFrame([
    {
        "posicao": posicao,
        "perfis": len(X),
        "variaveis": X.shape[1],
        "valores_ausentes": X.isna().sum().sum(),
    }
    for posicao, X in X_por_posicao.items()
])

resumo_entradas
```

### Padronização das variáveis por posição

Padronizamos as métricas separadamente para cada posição utilizando
o StandardScaler.

Cada valor é transformado subtraindo a média da variável e dividindo
pelo seu desvio-padrão. Nas variáveis não constantes, o resultado possui
média aproximadamente zero e desvio-padrão igual a um.

A interpretação passa a ser relativa aos jogadores da mesma posição:

- **0:** valor igual à média.
- **+1:** um desvio-padrão acima da média.
- **-1:** um desvio-padrão abaixo da média.

Um valor maior não significa necessariamente melhor desempenho:
em mortes por minuto, por exemplo, representa maior frequência de mortes.

A padronização não transforma as distribuições em normais nem remove
valores extremos. Preservamos os dados originais para interpretar
os grupos posteriormente.

```python
scalers_por_posicao = {}
X_padronizado_por_posicao = {}

for posicao, X in X_por_posicao.items():
    if X.isna().any().any():
        raise ValueError(f"Existem valores ausentes na posição {posicao}.")

    scaler = StandardScaler()

    X_padronizado = pd.DataFrame(
        scaler.fit_transform(X),
        index=X.index,
        columns=X.columns,
    )

    scalers_por_posicao[posicao] = scaler
    X_padronizado_por_posicao[posicao] = X_padronizado

# Conferência das médias e desvios após a transformação.
verificacao_padronizacao = pd.concat(
    {
        posicao: pd.DataFrame({
            "media": X.mean(),
            "desvio_padrao": X.std(ddof=0),
        })
        for posicao, X in X_padronizado_por_posicao.items()
    },
    names=["posicao", "metrica"],
)

verificacao_padronizacao.round(6)
```

## Modelagem não supervisionada

### Avaliação da quantidade de grupos

Aplicamos K-Means às métricas padronizadas, separadamente por posição,
testando de 2 a 6 grupos.

Avaliamos cada configuração com:

- **Silhueta:** mede a coesão e separação dos grupos. Varia de -1 a 1;
  valores próximos de zero indicam sobreposição.
- **Inércia:** soma das distâncias quadráticas aos centros dos grupos.
  Diminui conforme aumentamos a quantidade de grupos.
- **Tamanho dos grupos:** permite identificar configurações que isolam
  poucos jogadores.

Utilizamos 20 inicializações por configuração e uma semente fixa para
reprodutibilidade.

O maior valor de silhueta indica uma configuração candidata, mas a escolha
final também depende da interpretação e da estabilidade dos grupos.
O K-Means produz divisões mesmo quando não existem grupos naturais claros.

```python
resultados_kmeans = []
modelos_testados = {}

for posicao, X in X_padronizado_por_posicao.items():
    for k in range(2, 7):
        modelo = KMeans(
            n_clusters=k,
            n_init=20,
            random_state=42,
        )

        rotulos = modelo.fit_predict(X)
        tamanhos = pd.Series(rotulos).value_counts()

        resultados_kmeans.append({
            "posicao": posicao,
            "k": k,
            "silhueta": silhouette_score(X, rotulos),
            "inercia": modelo.inertia_,
            "menor_grupo": tamanhos.min(),
            "maior_grupo": tamanhos.max(),
        })

        modelos_testados[(posicao, k)] = modelo

df_avaliacao_kmeans = pd.DataFrame(resultados_kmeans)

df_avaliacao_kmeans.round(3)
```

### Comparação das configurações

Visualizamos a silhueta e a inércia para cada quantidade de grupos.

Procuramos configurações com boa separação e avaliamos se o aumento
da quantidade de grupos traz ganhos relevantes. A menor inércia,
isoladamente, não determina a melhor configuração.

```python
fig, eixos = plt.subplots(1, 2, figsize=(14, 5))

for eixo, metrica, titulo in zip(
    eixos,
    ["silhueta", "inercia"],
    ["Silhueta por quantidade de grupos", "Inércia por quantidade de grupos"],
):
    sns.lineplot(
        data=df_avaliacao_kmeans,
        x="k",
        y=metrica,
        hue="posicao",
        marker="o",
        ax=eixo,
    )

    eixo.set_title(titulo)
    eixo.set_xlabel("Quantidade de grupos")
    eixo.set_ylabel(metrica.capitalize())
    eixo.set_xticks(range(2, 7))

plt.tight_layout()
plt.show()

# Melhor silhueta por posição — indicação inicial, não decisão definitiva.
candidatos_k = (
    df_avaliacao_kmeans
    .sort_values(
        ["posicao", "silhueta", "k"],
        ascending=[True, False, True],
    )
    .drop_duplicates("posicao")
    .reset_index(drop=True)
)

candidatos_k.round(3)
```

### Caracterização dos grupos

Utilizamos provisoriamente dois grupos por posição, configuração com
a maior silhueta entre as opções testadas.

Relacionamos os agrupamentos aos jogadores e calculamos as médias
das características em suas unidades originais.

Os números dos grupos são apenas identificadores: o grupo 0 não é melhor
nem pior que o grupo 1. Grupos de posições diferentes também não possuem
correspondência automática.

Como a silhueta indicou sobreposição, os agrupamentos serão interpretados
como divisões exploratórias, ainda sujeitas à avaliação de estabilidade.

```python
df_clusters = df_modelo.copy()
df_clusters["cluster"] = pd.Series(
    pd.NA,
    index=df_clusters.index,
    dtype="Int64",
)

modelos_por_posicao = {}

for posicao, X in X_padronizado_por_posicao.items():
    # Reutiliza o modelo já ajustado na comparação anterior.
    modelo = modelos_testados[(posicao, 2)]
    modelos_por_posicao[posicao] = modelo

    df_clusters.loc[X.index, "cluster"] = modelo.predict(X)

resumo_clusters = (
    df_clusters
    .groupby(["position", "cluster"])[features_modelo]
    .mean()
)

resumo_clusters.insert(
    0,
    "jogadores",
    df_clusters.groupby(["position", "cluster"]).size(),
)

resumo_clusters.round(3)
```

### Características dos grupos em relação à posição

Visualizamos as médias padronizadas de cada grupo.

- Valores positivos indicam uma média acima da referência da posição.
- Valores negativos indicam uma média abaixo dessa referência.
- Valores próximos de zero indicam proximidade com a média da posição.

As cores representam diferenças estatísticas, não qualidade.
Por exemplo, um valor positivo em mortes por minuto indica mais mortes.

Cada jogador tem o mesmo peso no cálculo da média do grupo.

```python
rotulos_metricas = {
    "dpm_media": "Dano/min",
    "earnedgoldshare_media": "Participação no ouro",
    "cspm_media": "CS/min",
    "wpm_media": "Sentinelas/min",
    "wcpm_media": "Sentinelas removidas/min",
    "deaths_pm_media": "Mortes/min",
    "assists_pm_media": "Assistências/min",
    "participacao_abates_media": "Participação em abates",
}

medias_padronizadas = {}

for posicao, X in X_padronizado_por_posicao.items():
    dados = X.copy()
    dados["cluster"] = df_clusters.loc[X.index, "cluster"]

    medias_padronizadas[posicao] = (
        dados.groupby("cluster")[features_modelo].mean()
    )

mapa_grupos = pd.concat(
    medias_padronizadas,
    names=["posicao", "cluster"],
).rename(columns=rotulos_metricas)

limite_cor = mapa_grupos.abs().max().max()

plt.figure(figsize=(13, 7))

sns.heatmap(
    mapa_grupos,
    annot=True,
    fmt=".2f",
    cmap="vlag",
    center=0,
    vmin=-limite_cor,
    vmax=limite_cor,
    cbar_kws={"label": "Média padronizada"},
)

plt.title("Características dos grupos em relação à própria posição")
plt.xlabel("Métrica")
plt.ylabel("Posição e grupo")
plt.tight_layout()
plt.show()
```

### Jogadores de cada grupo

Consultamos os integrantes de cada grupo para relacionar as diferenças

```python
posicao_consulta = "mid"

jogadores_por_grupo = (
    df_clusters.loc[
        df_clusters["position"].eq(posicao_consulta),
        ["jogador", "ligas", "equipes", "partidas", "cluster"],
    ]
    .sort_values(
        ["cluster", "partidas", "jogador"],
        ascending=[True, False, True],
    )
)

# Mostra todos os jogadores da posição dentro deste bloco.
with pd.option_context("display.max_rows", None):
    display(jogadores_por_grupo)
```

```python
from itertools import combinations
from sklearn.metrics import adjusted_rand_score
```

### Estabilidade dos agrupamentos entre inicializações

Repetimos o K-Means com dois grupos e dez sementes diferentes, mantendo
os mesmos jogadores, variáveis e dados padronizados.

Cada execução utiliza 20 inicializações internas. Comparamos todos os
pares de resultados pelo Índice de Rand Ajustado (ARI).

Valores próximos de 1 indicam que a divisão se repete entre execuções.
A comparação independe da numeração dos grupos.

Este teste avalia apenas a sensibilidade à inicialização. Não demonstra
estabilidade diante de novas partidas, alterações nas variáveis ou
mudanças no conjunto de jogadores. Uma divisão estável também pode
apresentar baixa separação entre os grupos.

```python
sementes = range(10)

resultados_estabilidade = []
comparacoes_ari = []

for posicao, X in X_padronizado_por_posicao.items():
    rotulos_por_semente = {}

    for semente in sementes:
        modelo = KMeans(
            n_clusters=2,
            n_init=20,
            random_state=semente,
        )

        rotulos_por_semente[semente] = modelo.fit_predict(X)

    valores_ari = []

    for semente_a, semente_b in combinations(sementes, 2):
        ari = adjusted_rand_score(
            rotulos_por_semente[semente_a],
            rotulos_por_semente[semente_b],
        )

        valores_ari.append(ari)

        comparacoes_ari.append({
            "posicao": posicao,
            "semente_a": semente_a,
            "semente_b": semente_b,
            "ari": ari,
        })

    resultados_estabilidade.append({
        "posicao": posicao,
        "comparacoes": len(valores_ari),
        "ari_medio": pd.Series(valores_ari).mean(),
        "ari_minimo": min(valores_ari),
        "ari_maximo": max(valores_ari),
    })

df_comparacoes_ari = pd.DataFrame(comparacoes_ari)

df_estabilidade = (
    pd.DataFrame(resultados_estabilidade)
    .sort_values("ari_medio", ascending=False)
    .reset_index(drop=True)
)

df_estabilidade.round(3)
```

### Resultado da avaliação de estabilidade

Os agrupamentos apresentaram elevada estabilidade entre as dez sementes
testadas, mantendo 20 inicializações internas por execução.

Bot, jungle, mid e top apresentaram ARI igual a 1 nas comparações,
considerando a precisão exibida. Suporte apresentou ARI médio de 0,979
e mínimo de 0,94, indicando pequenas diferenças entre execuções.

Em conjunto com as silhuetas relativamente baixas, os resultados apontam
para divisões repetíveis, mas com sobreposição entre os perfis.

Essa conclusão se restringe à inicialização do algoritmo sobre a mesma

### Comparação de agrupamentos com maior detalhamento

A configuração com dois grupos apresentou a maior silhueta entre as
alternativas testadas, mas produziu divisões amplas, principalmente
relacionadas a recursos e desempenho observado.

Para investigar perfis mais específicos, compararemos configurações
com 2, 3 e 4 grupos em cada posição.

Aumentar o número de grupos permite maior detalhamento, mas também pode
fragmentar perfis semelhantes ou criar grupos pequenos. Portanto,
avaliaremos:

- Características médias dos grupos.
- Quantidade de jogadores em cada grupo.
- Silhueta e estabilidade entre inicializações.
- Utilidade das diferenças encontradas para interpretar os perfis.

Manteremos os mesmos jogadores, variáveis e padronização para que a
comparação reflita apenas a mudança no número de grupos.

As configurações serão exploratórias. Não presumimos que mais grupos
sejam melhores, e cada posição poderá utilizar uma quantidade diferente.

```python
# Depois, altere para "top", "jng", "bot" ou "sup".
posicao_comparacao = "mid"
valores_k = [2, 3, 4]

X = X_padronizado_por_posicao[posicao_comparacao]

perfis_por_k = {}
tamanhos_por_k = {}

for k in valores_k:
    modelo = modelos_testados[(posicao_comparacao, k)]
    rotulos = pd.Series(
        modelo.predict(X),
        index=X.index,
        name="cluster",
    )

    perfis_por_k[k] = (
        X.assign(cluster=rotulos)
        .groupby("cluster")
        .mean()
    )

    tamanhos_por_k[k] = rotulos.value_counts().sort_index()

# Mesma escala de cores em todos os gráficos.
limite_cor = max(
    tabela.abs().max().max()
    for tabela in perfis_por_k.values()
)

fig, eixos = plt.subplots(3, 1, figsize=(13, 12))

for eixo, k in zip(eixos, valores_k):
    tabela = perfis_por_k[k].rename(columns=rotulos_metricas)

    nomes_grupos = [
        f"Grupo {grupo} (n={tamanhos_por_k[k].loc[grupo]})"
        for grupo in tabela.index
    ]

    sns.heatmap(
        tabela,
        annot=True,
        fmt=".2f",
        cmap="vlag",
        center=0,
        vmin=-limite_cor,
        vmax=limite_cor,
        yticklabels=nomes_grupos,
        ax=eixo,
    )

    eixo.set_title(f"{posicao_comparacao} — {k} grupos")
    eixo.set_xlabel("")
    eixo.set_ylabel("")
    eixo.tick_params(axis="y", rotation=0)

plt.tight_layout()
plt.show()

# Indicadores das configurações exibidas.
df_avaliacao_kmeans.loc[
    df_avaliacao_kmeans["posicao"].eq(posicao_comparacao)
    & df_avaliacao_kmeans["k"].isin(valores_k)
].sort_values("k").round(3)
```

### Estabilidade das configurações com maior detalhamento

A comparação dos perfis de mid mostrou diferenças adicionais com três
e quatro grupos. Agora avaliamos se essas divisões se repetem quando
alteramos a inicialização do K-Means.

Testamos 2, 3 e 4 grupos em cada posição, utilizando dez sementes e
20 inicializações internas por execução.

Calculamos o ARI entre todos os pares de execuções e reunimos os resultados
com a silhueta e os tamanhos dos grupos do modelo de referência
(random_state=42).

A escolha considerará conjuntamente interpretação, separação, estabilidade
e tamanho dos grupos. Este teste avalia apenas a sensibilidade à
inicialização sobre os mesmos dados.

```python
sementes_teste = range(10)
valores_k_teste = [2, 3, 4]

resultados_estabilidade_k = []

for posicao, X in X_padronizado_por_posicao.items():
    for k in valores_k_teste:
        rotulos_execucoes = []

        for semente in sementes_teste:
            modelo = KMeans(
                n_clusters=k,
                n_init=20,
                random_state=semente,
            )

            rotulos_execucoes.append(modelo.fit_predict(X))

        valores_ari = [
            adjusted_rand_score(rotulos_a, rotulos_b)
            for rotulos_a, rotulos_b
            in combinations(rotulos_execucoes, 2)
        ]

        # Compara também as execuções com o modelo interpretado nos gráficos.
        rotulos_referencia = modelos_testados[
            (posicao, k)
        ].predict(X)

        ari_referencia = [
            adjusted_rand_score(rotulos_referencia, rotulos)
            for rotulos in rotulos_execucoes
        ]

        resultados_estabilidade_k.append({
            "posicao": posicao,
            "k": k,
            "ari_medio": pd.Series(valores_ari).mean(),
            "ari_minimo": min(valores_ari),
            "ari_referencia_medio": pd.Series(ari_referencia).mean(),
        })

df_estabilidade_k = pd.DataFrame(resultados_estabilidade_k)

df_comparacao_k = (
    df_avaliacao_kmeans
    .merge(
        df_estabilidade_k,
        on=["posicao", "k"],
        how="inner",
        validate="one_to_one",
    )
    .sort_values(["posicao", "k"])
    .reset_index(drop=True)
)

df_comparacao_k[
    [
        "posicao",
        "k",
        "silhueta",
        "menor_grupo",
        "maior_grupo",
        "ari_medio",
        "ari_minimo",
        "ari_referencia_medio",
    ]
].round(3)
```

### Busca de jogadores com perfis semelhantes

Calculamos a distância euclidiana entre um jogador de referência e os
demais jogadores da mesma posição, utilizando as oito métricas padronizadas.

Quanto menor a distância, maior a semelhança estatística nesse conjunto
de características. A consulta considera todos os jogadores elegíveis da
posição, independentemente do grupo atribuído pelo K-Means.

O próprio jogador é excluído dos resultados. A distância não representa
probabilidade, porcentagem de semelhança ou equivalência de habilidade.

A busca utiliza os perfis com pelo menos 20 partidas. As comparações
refletem o período analisado e o contexto das equipes, ligas e campeões.

```python
def buscar_jogadores_semelhantes(nome, posicao="mid", quantidade=5):
    if posicao not in X_padronizado_por_posicao:
        raise ValueError(f"Posição inválida: {posicao}")

    if quantidade < 1:
        raise ValueError("A quantidade deve ser maior que zero.")

    X = X_padronizado_por_posicao[posicao]
    perfis = df_modelo.loc[X.index]

    encontrados = perfis.loc[
        perfis["jogador"].str.casefold().eq(nome.strip().casefold())
    ]

    if encontrados.empty:
        raise ValueError(
            f"Jogador '{nome}' não encontrado na posição '{posicao}' "
            "entre os perfis com pelo menos 20 partidas."
        )

    if len(encontrados) > 1:
        raise ValueError(
            "Existe mais de um jogador com esse nome na posição. "
            "Verifique os playerids antes de continuar."
        )

    indice_referencia = encontrados.index[0]
    referencia = X.loc[indice_referencia]

    # Raiz da soma das diferenças ao quadrado entre as métricas.
    distancias = (
        X.sub(referencia)
        .pow(2)
        .sum(axis=1)
        .pow(0.5)
        .drop(index=indice_referencia)
        .sort_values()
        .head(quantidade)
        .rename("distancia")
    )

    resultado = perfis.loc[
        distancias.index,
        ["playerid", "jogador", "position", "ligas", "equipes", "partidas"],
    ].copy()

    resultado["distancia"] = distancias
    resultado.insert(0, "ordem", range(1, len(resultado) + 1))

    return resultado
```

```python
jogador_referencia = "Faker"
posicao_consulta = "mid"

jogadores_semelhantes = buscar_jogadores_semelhantes(
    nome=jogador_referencia,
    posicao=posicao_consulta,
    quantidade=5,
)

jogadores_semelhantes.round({"distancia": 3})
```

### Relação dos grupos com vitórias e equipes

Calculamos a taxa de vitória de cada jogador na posição analisada e
comparamos sua distribuição entre os grupos do modelo com k = 2.

Cada jogador tem o mesmo peso na média do grupo, independentemente da
quantidade de partidas. O resultado das partidas não participou do
treinamento e é utilizado apenas para interpretação.

Também verificamos a distribuição dos jogadores entre grupos dentro
de cada equipe e posição. Jogadores que trocaram de equipe podem aparecer
em mais de uma linha dessa análise, mantendo o grupo de seu perfil agregado.

As relações observadas são descritivas: não permitem separar os efeitos
do jogador, da equipe, dos adversários e dos campeões utilizados.

```python
# Recupera os grupos de referência com k = 2.
df_contexto_clusters = df_modelo.copy()

df_contexto_clusters["cluster"] = pd.Series(
    pd.NA,
    index=df_contexto_clusters.index,
    dtype="Int64",
)

for posicao, X in X_padronizado_por_posicao.items():
    df_contexto_clusters.loc[X.index, "cluster"] = (
        modelos_testados[(posicao, 2)].predict(X)
    )

# Taxa de vitória individual no período e na posição.
taxas_vitoria = (
    df_jogadores_limpo
    .groupby(["playerid", "position"])["result"]
    .mean()
    .mul(100)
    .rename("vitorias_pct")
    .reset_index()
)

df_contexto_clusters = df_contexto_clusters.merge(
    taxas_vitoria,
    on=["playerid", "position"],
    how="left",
    validate="one_to_one",
)

resumo_vitorias = (
    df_contexto_clusters
    .groupby(["position", "cluster"])
    .agg(
        jogadores=("playerid", "size"),
        vitorias_media_pct=("vitorias_pct", "mean"),
        vitorias_mediana_pct=("vitorias_pct", "median"),
    )
)

display(resumo_vitorias.round(2))

plt.figure(figsize=(11, 5))

sns.boxplot(
    data=df_contexto_clusters,
    x="position",
    y="vitorias_pct",
    hue="cluster",
    order=["top", "jng", "mid", "bot", "sup"],
)

plt.title("Taxa de vitória dos jogadores por posição e grupo")
plt.xlabel("Posição")
plt.ylabel("Vitórias (%)")
plt.tight_layout()
plt.show()

# Vincula cada jogador às equipes em que efetivamente atuou.
vinculos_equipes = df_jogadores_limpo[
    ["playerid", "position", "league", "teamid", "teamname"]
].drop_duplicates()

vinculos_equipes = vinculos_equipes.merge(
    df_contexto_clusters[["playerid", "position", "cluster"]],
    on=["playerid", "position"],
    how="inner",
    validate="many_to_one",
)

distribuicao_equipes = pd.crosstab(
    [
        vinculos_equipes["position"],
        vinculos_equipes["league"],
        vinculos_equipes["teamname"],
    ],
    vinculos_equipes["cluster"],
)

# Altere a posição para examinar as demais.
display(distribuicao_equipes.loc["mid"])
```
